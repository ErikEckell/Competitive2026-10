# Memoization

Memoization is an optimization technique used primarily to speed up programs by storing the results of expensive calls to pure functions, and returning the cached result when the same input occurs again. This is normaly used in Dynamic programming approaches.

## Example: Fibonacci

Fibonacci is a sequence of numbers that can be defined by the following equations:

$$fib_0 = 0$$
$$fib_1 = 1$$
$$fib_n = fib_{n-1} + fib_{n-2}$$


### Brute Force
A brute force implementation of the Fibonacci sequence:

```python
def fib_brute(n):
    fib = [0, 1]
    if n > 1:
        while len(fib) < n+1:
            fib.append(fib[-1] + fib[-2])
    return fib[n]
    
```

```python
fib_brute(8)
```

### Recursive

A recursive implementation of the Fibonacci sequence:

```python
def fib_rec(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fib_rec(n-1) + fib_rec(n-2)
```

```python
fib_rec(8)
```

### Memoization

Both of the previous implementations have the issue that the Fibonacci sequence must be computed for every independent execution, whichcan be expensive if multiple calls are made (specially with the recursive implementation). Both implementations can be improved with Memoization

#### Brute Force Memoization

```python
mem_brute = {
    0: 0,
    1: 1
}
def fib_brute_mem(n):
    if n not in mem_brute:
        latest_n = max(mem_brute.keys()) + 1
        while latest_n <= n:
            mem_brute[latest_n] = mem_brute[latest_n-1] + mem_brute[latest_n-2]
            latest_n += 1
    return mem_brute[n]
```

```python
mem_brute
```

```python
fib_brute_mem(8)
```

```python
mem_brute
```

#### Recursive Memoization

```python
mem_rec = {
    0: 0,
    1: 1
}
def fib_rec_mem(n):
    if n not in mem_rec:
        mem_rec[n] = fib_rec_mem(n-1) + fib_rec_mem(n-2)
    return mem_rec[n]
```

```python
mem_rec
```

```python
fib_rec_mem(8)
```

```python
mem_rec
```

### Comparissons

## Decorators

You can encapsulate your Memoization implementation with a custom class

```python
class Memoize:
    def __init__(self, f):
        self.f = f
        self.memo = {}
    def __call__(self, *args):
        if not args in self.memo:
            self.memo[args] = self.f(*args)
        return self.memo[args]
```

```python
@Memoize
def fib_rec_dec(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fib_rec(n-1) + fib_rec(n-2)
```

```python
fib_rec_dec(35)
```

```python
fib_rec_dec(35)
```
