## `For`

Syntax for the for loop is:
```python
for X in Y:
    CODE
[else:
    CODE]
```

````python
for i in range(5):
    print(i)
````

````python
# If the for isn't "broken", the else block of the
# for loop is executed
for i in range(5):
    print(i)
else:
    print("Clean finish")
````

````python
# If there is a break in the for loop, the else block
# isn't executed
for j in range(5):
    print(j)
    if j > 2:
        break
else:
    print("Clean finish")
````

To iterate over a list, don't:
```Python
items = list(...)
for i in range(len(items)):
    item = items[i]
    do_something(item)
```
Instead use:
```Python
for item in items:
    do_something(item)
```
Or if you need the index `i`:
```Python
for i, item in enumerate(item):
    do_something(item, i)
```

If you need to transform the list, do:
```Python
transform = list(map(transform, items))
```
instead of:
```Python
transform = []
for item in items:
    transform.append(transform(item))
```

## Iterators and Generators

Useful concepts in `Python` are **iterators** and **generators**. An iterator is an object for
which we can ask give us the `next()` item.

A generator (small lie) is a type of iterator, which is defined similar to a python function
but it does not necesarely return a single value, but yields multiple values.
Generators are defined by placing a `yield` statement where we want to yield a value and halt the generator execution.

````python
def count():
    current = 0
    while True:
        current += 1
        yield current
````

````python
count()
````

````python
c = count()
for i in range(5):
    print(next(c))
````

Generators allows us to define lazy arrays. An array that could contain all objects I need
(and even a conceptually infinite number of objects), but only creates them as required.

Example:

A generator for all the prime numbers

````python
def primes():
    i = 3
    prime_list = [2]
    yield 2
    while True:
        for p in prime_list:
            if i % p == 0:
                break
            else:
                prime_list.append(i)
            yield i
            i += 2
````

````python
k = 0
for p in primes():
    print("The {0}-th prime is '{1}'".format(k, p))
    if k > 4:
        break
    k += 1
````

## Working with Iterators ( `map` - `filter` - `zip` and `enumerate` )

`Python` provides functionalities to work directly with iterators. Useful examples are the `map` and `filter` functions.

To use them, it is common to use lambda expression to avoid writing functions definitions that are only used once. A lambda expression uses the keyword `lambda` to define on-the-fly functions.

````python
addition = lambda x,y: x+y
addition(1,2)
````

NOTE: The previously defined addition lambda expression, is equivalent to:

````python
def addition(x,y):
    return x + y
````

`map` applies a function to all the elements of the iterator:

```Python
[f(x1), ..., f(xn)] = map(f, [x1,...,xn])
```

`filter` fitlers the elements of an iterator that evaluate the function to `True`

```Python
[x_i1, x_i2, ..., x_iK] = filter(f, [x_1,...,x_N])
```

NOTE: both `map` and `filter` return an `iterator`.

````python
sq = map(lambda x: x**2, range(5))
print(sq)
````

````python
for i in sq:
    print(i)
````

````python
even = filter(lambda x: x%2 == 0, range(10))
for i in even:
    print(i)
````

Although `filter` can be replaced with list comprehesion's conditionals:
```Python
even = (x for x in range(20) if x % 2 == 0)
```
Other useful iterators' utilities are `enumerate` and `zip`.

`enumerate` yields the same elements as the arguments iterator, but also yields a counter of how many elements have been extracted from the iterator:
```Python
for (i, elem) in enumerate(iterator):
...
```

`zip` merges two (or more) iterators and return a tuple with the i-th elements of each iterator.
```Python
for elem1, elem2 in zip(iterador1, iterador2):
...
```

````python
a = ['Hello', 'Bye', None, 3, 2]
for i, elem in enumerate(a):
    print("The {0}-th element is '{1}'".format(i, elem))
````

````python
b = ["World", "Bye", "of the above", 1, 2, 3, 2, 1]
for e1,e2 in zip(a,b):
    print(f"{e1} - {e2}")
````
