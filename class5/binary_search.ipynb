# Binary Search

## Classical

Classical binary search is a way to quickly search for valuen within an array. It works as follows:

Problem statement: given an sorted array of number `A` , find the first element greater than `x`

```Python
# pseudocode
def binary_search(arr, x):
    a = 0
    b = len(arr) - 1
    while a < b:
        c = a + (b - a) / 2
        if arr[c] >= x:
            b = c
        else:
            a = c + 1

    if arr[a] < x:
        return None
    return arr[a] # or arr[b]
```

### Using Python

Python provides `bisect` library which implements binary search for us.

```Python
bisect.bisect_left(a, x, lo=0, hi=len(a), *, key=None)
bisect.bisect(a, x, lo=0, hi=len(a), *, key=None)
```

As an example:

````python
from bisect import bisect
def grade(score):
    breakpoints = [60, 70, 80, 90]
    grades = 'FDCBA'
    i = bisect(breakpoints, score)
    return grades[i]
````

````python
grade(33)
````

````python
grade(80)
````

````python
grade(100)
````

## Lazy/Computed list

Given `bisect`'s implementation it may be beneficial to define a list that it is defined by a function:

````python
from functools import cache # python >= 3.9

class lazy_list:
    def __init__(self, f):
        self.f = f
    
    @cache
    def __getitem__(self, key):
        return self.f(key)
````

We can do an implementation of `bisect` with this memoization technique:

````python
lazy_bisect = lazy_list(bisect)
````

## Generalization

This algorithm can be extended as follows:

- A linearly sorted domain $A$ (for all elements $x, y$ ; they either are $x < y$ or $y < x$ )
- A property `P(x)` with the property of:
    - If `P(x) = True` then for all $y > x$ , `P(y)` is `True`
    - If `P(x) = False` then for all $y < x$ , `P(y)` is `False`


Binary search can be extended to a more abstract algorithm to find the first element $x$ of the domain `arr` with the property `P(x)` as follows:

```Python
def general_binary_search_array(arr, P):
    a = 0
    b = len(A) - 1
    while a < b:
        c = a + (b - a) / 2
        if P(arr[c]):
            b = c
        else:
            a = c + 1
    if not P(arr[a]):
        return None
    return arr[a] # or arr[b]
```

The same algorithm can be achieved with `bisect` by using the key parameter and looking for the
first `True` value.

## Continuous case

This general approach can be extended to work with continuous domains as follows:

```Python
def general_binary_search_continuous(
    f: Callable[[float], bool],
    lo: float,
    hi: float,
    eps: float
) -> float:
    while hi - lo > eps:
        mid = lo + (hi - lo) / 2
        if f(mid):
            hi = mid
        else:
            lo = mid
    return lo # or "return hi"
```

## Example

Peter received money from his parents this week and wants to spend it all buying books. But he does not read a book so fast, because he likes to enjoy every single word while he is reading. In this way, it takes him a week to finish a book.

As Peter receives money every two weeks, he decided to buy two books, then he can read them until
receive more money. As he wishes to spend all the money, he should choose two books whose prices
summed up are equal to the money that he has. It is a little bit difficult to find these books, so Peter asks your help to find them.

````python
from bisect import bisect_left
from itertools import zip_longest

def calculate(money, book_list):
    books = sorted(book_list)
    best = (None, None)
    num_books = len(books)
    for i, b1 in enumerate(books[:num_books//2+1]):
        j = bisect_left(books[i+1:], money - b1)
        if j == num_books - i - 1 or (b2 := books[j+i+1] ) + b1 > money: continue
        if best[0] is None or (b2 - b1 < best[1] - best[0]):
            best = (b1, b2)
    print(f"Peter should buy books whose prices are {best[0]} and {best[1]}.")
````
